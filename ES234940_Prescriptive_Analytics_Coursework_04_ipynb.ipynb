**Solution**

a. Mathematical Model:

<p>
This problem can be formulated as a <strong>0-1 Knapsack Problem</strong>.
The traveler must select tourist attractions to maximize the total satisfaction score without exceeding the available budget.
</p>

<h4>1. Decision Variables</h4>

<p>
Let:
</p>

<p>
x<sub>i</sub> =
</p>

<ul>
  <li>1, if tourist attraction <em>i</em> is selected</li>
  <li>0, if tourist attraction <em>i</em> is not selected</li>
</ul>

<p>
where <em>i = 1, 2, 3, ..., 8</em>.
</p>

<h4>2. Parameters</h4>

<p>
Let:
</p>

<ul>
  <li><em>c<sub>i</sub></em> = cost of tourist attraction <em>i</em></li>
  <li><em>s<sub>i</sub></em> = satisfaction score of tourist attraction <em>i</em></li>
  <li><em>B</em> = available budget = Rp. 1,000,000</li>
</ul>

<h4>3. Hard Constraint</h4>

<p>
The total cost of selected tourist attractions must not exceed the available budget.
</p>

<p>
&sum;<sub>i=1</sub><sup>8</sup> c<sub>i</sub>x<sub>i</sub> &le; B
</p>

<p>
Since each tourist attraction is either selected or not selected:
</p>

<p>
x<sub>i</sub> &isin; {0, 1}, for i = 1, 2, 3, ..., 8
</p>

<h4>4. Objective Function</h4>

<p>
The objective is to maximize the total satisfaction score.
</p>

<p>
Maximize:
</p>

<p>
Z = &sum;<sub>i=1</sub><sup>8</sup> s<sub>i</sub>x<sub>i</sub>
</p>

```python
from ortools.sat.python import cp_model

# Data for the problem (bids)
costs = [
    [15, 10, 9],  # John's bids for Mow, Paint, Wash
    [9, 15, 10],  # Karen's bids for Mow, Paint, Wash
    [10, 12, 8]   # Terri's bids for Mow, Paint, Wash
]

num_children = len(costs)
num_chores = len(costs[0])

# Create the CP-SAT model
model = cp_model.CpModel()

# Decision variables
# x[i, j] is 1 if child i is assigned to chore j, and 0 otherwise.
x = {}
for i in range(num_children):
    for j in range(num_chores):
        x[i, j] = model.NewBoolVar(f'x_{i}_{j}')

# Constraints
# Each child is assigned exactly one chore.
for i in range(num_children):
    model.AddExactlyOne(x[i, j] for j in range(num_chores))

# Each chore is assigned to exactly one child.
for j in range(num_chores):
    model.AddExactlyOne(x[i, j] for i in range(num_children))

# Objective function: Minimize total cost
model.Minimize(sum(costs[i][j] * x[i, j] for i in range(num_children) for j in range(num_chores)))

# Solve the model
solver = cp_model.CpSolver()
solver.parameters.log_search_progress = True
status = solver.Solve(model)

# Print the solution
if status == cp_model.OPTIMAL or status == cp_model.FEASIBLE:
    print(f'Total cost = {solver.ObjectiveValue()}')
    print('Assignments:')
    children = ['John', 'Karen', 'Terri']
    chores = ['Mow', 'Paint', 'Wash']
    for i in range(num_children):
        for j in range(num_chores):
            if solver.Value(x[i, j]) == 1:
                print(f'  {children[i]} assigned to {chores[j]} (cost: ${costs[i][j]})')
else:
    print('No solution found.')

print(f'\nSolver statistics:')
print(f'  - conflicts      : {solver.NumConflicts()}')
print(f'  - branches       : {solver.NumBranches()}')
print(f'  - wall time      : {solver.WallTime()} s')
```
